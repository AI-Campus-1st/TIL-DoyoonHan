# 문항 1 VGGNet 논문리뷰

## Deep Networks

* Fixed other parameters of the architecture, and only increased the depth of the network by adding more convolutional layers. (depth of 11~19)

## Small Filters, Max PoolingSimple Architecture

* Used very small (3x3) convolution filters in all layers.

* Fixed-size 224 x 224 RGB image.

* Only pre-processing done was subtracting the mean RGB value on the training set from each pixel

* Max-pooling performed over 2x2 window with stride 2.

* Allowed the decision function to be more discriminative and decreased the number of parameters

## ReLU Activation Function

* All hidden layers are equipped with ReLU function.

* None of the networks contain Local Response Normalisation as it does not imporve the performance but leads to increased memory consumption and computation time.

## Increased Non-Linearity

* Multiple 3x3 led to multiple ReLU function which resulted in increase in non-linearity.

## Effect of Depth

* Incrased depth of ConvNets led to better performance

## ILSVRC Performance

* Resulting Ensemble of 7 networks resulted to 7.3% test error

* Ensemble of only two best performing models reduced the test error to 7.0% using dense evaluation

* 6.8% using combined dense and multi-crop evaluation.

* best-performing single model achieved 7.1% error

# 문항 2 ResNet 논문리뷰

## Residual Learning

* Residual learning explicitly reformulates stacked layers to approximate a residual function relative to the inputs rather than learning an unreferenced mapping, making deep networks significantly easier to optimize and solving the degradation problem

## Skip Connection

* Skip connections implement residual learning by performing parameter-free identity mappings that pass input signals directly to layer outputs via element-wise addition, adding no extra computational complexity while enabling end-to-end training via standard backpropagation

## Deep Networks

* Driven by residual representations, deep networks can scale up to 152 layers on ImageNet with lower FLOP complexity than VGG nets, achieving accuracy gains with increased depth and securing 1st place across all major tasks in the ILSVRC & COCO 2015 competitions


# 문항 3 LSTM 논문리뷰

## Long-Term Memory

* Slowly changing weights

* The architecture is local in both space and time, requiring an update computational complexity per time step and weight of only O(1)

## Gates

* Multiplicative gate units learn to open and close access to the constant error flow within memory cells

* Input gate: Resolves the "input weight conflict" by controlling when new information is written to or kept in the memory cell, protecting stored contents from being corrupted by irrelevant incoming inputs

* Output gate: Resolves the "output weight conflict" by controlling when memory cell contents are read or accessed, preventing currently irrelevant stored information from disturbing other units in the network

## Gradient Vanishing

* In standard BPTT and RTRL algorithms, backpropagated error signals exponentially decay (vanish) or blow up over time, depending on weight magnitudes and activation derivatives

* LSTM enforces constant error flow through internal states

* truncates error gradients at specific points outside the CECs where they would otherwise decay or cause harm
