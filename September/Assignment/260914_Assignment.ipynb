# 선형회귀 모델 직접 구현

```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt

data = make_regression(n_samples = 1000, n_features = 1, bias = 200, noise = 30) 

X = data[0]
y = data[1]

n_data = len(X)
```

```python
# hypothesis: y = w * X + b

w = np.random.randn(1)
b = np.random.randn(1)

print("w: ", w)
print("b: ", b)
```

```python
epochs = 1000
learning_rate = 0.01

history = {
    'cost' : []
}

for epoch in range(epochs):
    y_pred = w * X + b

    cost = np.sum((y - y_pred.ravel()) ** 2)/n_data
    history['cost'].append(cost)

    dw1 = 2 * (y_pred.ravel() - y) @ (X / n_data)
    dw0 = 2 * np.mean((y_pred.ravel() - y))

    w = w - learning_rate * dw1
    b = b - learning_rate * dw0

plt.plot(history['cost'])
plt.xlabel('Epochs')
plt.ylabel('Cost')
plt.show()
```

```python
print("w: ", w)
print("b: ", b)
```

```python
final_pred = X @ w + b
error = final_pred - y

mse = np.mean(error ** 2)
rmse = np.sqrt(mse)

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
```

# 다변량 선형회귀 모델 직접 구현
