```python
from collections import Counter
import numpy as np
import pandas as pd
from IPython.display import display

stopwords = {"등", "것", "수", "때", "경우", "관련", "이번", "지난"}

keep_pos = {"NNG", "NNP", "SL"}

def preprocess(text):
    if not isinstance(text, str):
        raise TypeError("기사 본문은 문자열이어야 합니다.")

    text = " ".join(text.split())

    if not text:
        return []

    return [
        word
        for word, pos in komoran.pos(text)
        if pos in keep_pos and word not in stopwords
    ]

# 각 원소가 기사 하나의 단어 목록
docs_tokens = [
    preprocess(article["text"])
    for article in news
]

doc_ids = [f"D{i:02d}" for i in range(len(news))]

doc_info = pd.DataFrame({
    "문서": doc_ids,
    "제목": [article["title"] for article in news],
    "토큰 수": [len(words) for words in docs_tokens],
})

display(doc_info)
```

```python
def make_tdm(documents, document_ids):
    # 전체 문서에서 사용된 고유한 단어
    vocab = sorted({
        word
        for words in documents
        for word in words
    })

    word2id = {word: i for i, word in enumerate(vocab)}

    matrix = np.zeros(
        (len(vocab), len(documents)),
        dtype=int
    )

    for doc_idx, words in enumerate(documents):
        counts = Counter(words)

        for word, count in counts.items():
            matrix[word2id[word], doc_idx] = count

    return pd.DataFrame(
        matrix,
        index=pd.Index(vocab, name="단어"),
        columns=document_ids
    )

df_TDM = make_tdm(docs_tokens, doc_ids)
df_TDM
```

* 문서 수가 고정되어 있을 때 TDM의 크기를 줄이려면 단어의 종류를 줄여야합니다.

```python
tf = df_TDM.astype(float)
df = (df_TDM > 0).sum(axis=1)
N = df_TDM.shape[1]
idf = np.log(N / df)
df_TFIDF = tf.mul(idf, axis=0)

for doc_id, article in zip(doc_ids, news):
    scores = df_TFIDF[doc_id]

    top10 = (
        scores[scores > 0]
        .sort_values(ascending=False)
        .head(10)
    )

    print(f"\n[{doc_id}] {article['title']}")
    display(top10.rename("TF-IDF").to_frame())
```

* TDM은 여러 문서에 공통으로 자주 등장하는 단어도 높게 평가하지만 TF-IDF는 특정 문서에서 많이 나오는 단어들을 강조하는 성질을 가지고 있기 때문에 문서별로 분류하기 더 좋습니다.

```python
def make_ngrams(words, n):
    return [
        tuple(words[i:i + n])
        for i in range(len(words) - n + 1)
    ]
```

```python
docs_ngram_features = []

for words in docs_tokens:
    features = []

    for n in (1, 2, 3):
        for gram in make_ngrams(words, n):
            # 튜플 표현으로 토큰 경계도 보존
            features.append(f"{n}:{gram!r}")

    docs_ngram_features.append(features)

df_TDM_ngram = make_tdm(
    docs_ngram_features,
    doc_ids
)

print("1-gram TDM:", df_TDM.shape)
print("1~3-gram TDM:", df_TDM_ngram.shape)

top_features = (
    df_TDM_ngram.sum(axis=1)
    .sort_values(ascending=False)
    .head(20)
    .index
)

display(df_TDM_ngram.loc[top_features])
```

* 고려해야 할 것은 분석 목적과 분야 입니다.

* n이 너무 크면 n-gram이 unique할 확률이 높아 등장수가 낮을 확률이 높아집니다.

```python
def make_cooccurrence(words, window_size=2):
    vocab = sorted(set(words))
    word2id = {word: i for i, word in enumerate(vocab)}

    matrix = np.zeros(
        (len(vocab), len(vocab)),
        dtype=int
    )

    for i, center_word in enumerate(words):
        start = max(0, i - window_size)
        end = min(len(words), i + window_size + 1)

        for j in range(start, end):
            if i == j:
                continue

            context_word = words[j]

            matrix[
                word2id[center_word],
                word2id[context_word]
            ] += 1

    return pd.DataFrame(
        matrix,
        index=pd.Index(vocab, name="중심 단어"),
        columns=pd.Index(vocab, name="문맥 단어")
    )

selected_tokens = docs_tokens[7]

df_cooccurrence = make_cooccurrence(
    selected_tokens,
    window_size=2
)

print(news[7]["title"])

frequent_words = [
    word
    for word, count in Counter(selected_tokens).most_common(15)
]

display(
    df_cooccurrence.loc[frequent_words, frequent_words]
)
```

* 고려해야 할 것은 문장 길이와 분석 목적입니다. window_size가 작으면 전체적인 맥락을 놓칠수도 있지만 변화에 민감해 집니다. 반대로 크면 노이즈에 강해지지만 이와 동시에 불필요한 노이즈가 섞일수도 있습니다.
