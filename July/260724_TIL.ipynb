### Set

mutable, unordered, iterable

* .add()

* .update() 여러 값 추가

* .remove(elem) <- cause KeyError when elem not in set

* .discard(elem) <- does not cause error when elem not in set

* .pop() <- remove *random* element


### Dictionary

mutable, unordered, iterable

uses Key & Value pair.

* .keys()

* .values()

* .items()

* .get(key)

* .pop(key)

* .update() -> .update({"key" : value})
