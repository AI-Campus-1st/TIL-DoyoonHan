### Hash

hashlib library를 통해 hash 할 수 있고 여러가지 hash 알고리즘이 있다. (ex. SHA256)

hash로 바꾸는것은 쉬우나 hash에서 다시 바꾸는것은 매우 어려움. 그냥 password대신에 hash를 비교해서 password가 맞는지 비교.

hash를 바꾸는게 어려운 이유: hash의 알고리즘안에있는 알고리즘들이 어떻게 만들어 졌는지 모르기 때문에 그 사이에있는 프로세스를 revert하는것은 거의 불가능.

salt: hash algorithm에 user 마다 다른 salt값을 부여해 hash algorithm을 더욱 뚫기 어렵게 한다. hash 함수를 아는것만으로 뚫을수 없다. 하지만 db에 salt값이 저장이 되어있기 때문에 pepper도 사용한다

pepper: pepper는 db에 저장되어있지 않음. 보통 서버별로 하나의 pepper가 있음. 결과적으로 H(pw, salt, pepper)를 사용해 뚫는것을 막음.

* hash.new() 같은 것으로 여러 알고리즘사용가능

* hash.update() -> hash객체로 업데이트

* hash.digest() hash.hexdigest() -> update()로 업데이트된 객체를 요약반환. digest()는 byte로 반환, hexdigest()는 str로 반환


#### Hash와 Encryption(암호화)의 차이

Hash는 단방향, Encryption은 양방향.

Encryption 은 decrypt 할수있음. Hash는 불가능.
