# 문항 1 정중앙 문자

```python
def get_middle_char(word: str):
    length = len(word)
    if length % 2 == 0:
        return word[round(length/2) - 1] + word[round(length/2)]
    else:
        return word[round((length/2) - 0.5)]
```

* length//2 사용하면 바로 middle index 사용가능.

# 문항 2 가변 인자 리스트

```python
def my_avg(*args):
    return sum(args)/len(args)
```
