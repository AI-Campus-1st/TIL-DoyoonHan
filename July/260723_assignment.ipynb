### 문항 1 모음은 몇 개나 있을까?

```python
def vowel_counter(word: str):
    vowels = ['a', 'e', 'i', 'o', 'u']
    count = 0
    for character in word:
        if character.lower() in vowels:
            count += 1

    return count
```

### 문항 2 네모 출력

```python
def star_square(n: int, m: int):
    string = ((("*" * n) + "\n") * (m - 1)) + ("*" * n)

    return string

n = int(input("가로의 길이는? "))
m = int(input("세로의 길이는? "))

print(star_square(n, m))
```

### 문항 3 Built-in 함수와 메서드

```python
a = [1, 3, 2, 4, 6, 5]

# sorted()
# sorted는 리스트 자체를 바꾸지 않고, 정렬된 리스트를 return 한다.

print(sorted(a))
# [1, 2, 3, 4, 5, 6]

print(a)
# [1, 3, 2, 4, 6, 5]


# .sort()
# sort를 사용하면 리스트가 바뀐다.

a.sort()

# .sort()는 언제나 None을 return 하기 때문에 a.sort()를 print하면 안된다.
print(a.sort())
# None

print(a)
# [1, 2, 3, 4, 5, 6]
```
