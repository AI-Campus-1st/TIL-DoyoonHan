## Error & Exception

* try-except를 사용해 error handle

except as e 를 사용해 에러메시지 커스터마이징 가능

* else: 에러가 나오지 않았을 때 else안에있는것 사용

* finally를 사용해 반드시 수행해야하는 문장 입력가능 

* raise 사용해 error 불러올 수 있음.

* assert로 불러올 수도 있음. assert 사용시 무조건 AssertionError가 나옴.
