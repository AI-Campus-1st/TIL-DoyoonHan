# 문항 1 CSV 다루기

```python
def average_calculator(file: str):
    with open(file, 'r') as f:
        columns = f.readline()
        lines = f.readlines()

    col_names = columns.rstrip().split(',')
    avg_scores = {}

    for line in lines:
        data = line.rstrip().split(',')
        name = data.pop(0)

        total = 0
        pointer = 1
        for score in data:
            try:
                total += int(score)
            except:
                print(f"[처리] {name} - {col_names[pointer]} 값 오류({score}) → 0점 처리")
            finally:
                pointer += 1

        average = round(total/(len(columns) - 1), 1)
        avg_scores.update({name : average})

    with open("result.csv", 'w') as f:
        f.write("name,average\n")
        for key, value in avg_scores.items():
            f.write(key + ',' + str(value) + "\n")

    return print("완료: result.csv 생성")
```

```python
average_calculator("./scores.csv")
```

# 문항 2 CSV를 JSON으로 변환

```python
import json

def csv_to_json(file: str):
    with open(file, 'r') as f:
        columns = f.readline()
        lines = f.readlines()

    col_names = columns.rstrip().split(',')
    filename = file.rstrip(".csv")
    filename += ".json"

    result = []
    errors = []
    for line in lines:
        data = line.rstrip().split(',')
        id = data[0]
        name = data[1]
        age = data[2]

        try:
            # 이름이 비어있을 때
            if name == "":
                raise ValueError("name 비어있음")
            
            result.append({ col_names[0] : id, col_names[1] : name, col_names[2] : int(age)})
        except ValueError as e:
            errors.append(f"{int(id) + 1}행: {e}\n")

    # members.json creation
    with open("members.json", "w") as f:
        json.dump(result, f, ensure_ascii=False, indent=2)

    # errors.log creation
    with open("errors.log", "w") as f:
        for line in errors:
            f.write(line)
    
    return print(f"성공 {len(result)}건 / 실패 {len(errors)}건")
```

```python
csv_to_json("./members.csv")
```
