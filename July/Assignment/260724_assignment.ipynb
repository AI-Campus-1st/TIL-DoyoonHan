### 문항 1 복사가 잘 된 건가?

```python
a = [1, 2, 3, 4, 5]
b = a

a[2] = 5

print(a)
print(b)
```

### Answer

List a가 만들어졌고 b 에 a의 값을 부여했을 때, b 에 부여된것은 a의 메모리 위치이기 때문에 a[2] = 5로 list a의 element를 변경해도 b도 같이 변경된것처럼 보인다.

새로운 list를 만들고 싶으면 .copy()를 사용해 다른 메모리 주소에 새로운 list를 만들어야한다.

### 문항 2 평균 점수 구하기

```python
scores = {
    'python': 80,
    'algorithm': 90,
    'django': 89,
    'web': 83,
}

total = sum(scores.values())
mean = total / len(scores)
print(mean)
```

### 문항 3 혈액형 분류하기

```python
bloods = [
    'A', 'B', 'A', 'O', 'AB', 'AB',
    'O', 'A', 'B', 'O', 'B', 'AB',
]

def blood_dictionary(bloods):
    dictionary = {}
    for blood in bloods:
        if blood in dictionary:
            count = dictionary.get(blood)
            count += 1
            dictionary.update({blood : count})
        else:
            dictionary.update({blood : 1})

    return dictionary

print(blood_dictionary(bloods))
```
