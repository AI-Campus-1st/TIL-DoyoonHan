## 1. 숫자의 의미

```python
def get_secret_word(nums: list):
    word = ""
    for num in nums:
        if (num >= 65 and num <= 90) or (num >= 97 and num <= 122): 
            word += chr(num)
    return word

get_secret_word([83, 111, 115, 108, 110])
```

## 2. 내 이름은 몇일까?

```python
def get_secret_number(name: str):
    num = 0
    for character in name:
        num += ord(character)
    return num

get_secret_number('tom')
```

## 3. 강한 이름

```python
def get_strong_word(words: list):
    strong_word = ""
    i = 0
    for word in words:
        num = 0
        for character in word:
            num += ord(character)
        if num > i:
            strong_word = word
            i = num

    return strong_word

print(get_strong_word(['z', 'a']))
print(get_strong_word(['tom', 'john']))
```
