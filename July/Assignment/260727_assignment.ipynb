# 문항 1 홀수만 담기

```python
def odd():
    odd = []
    for i in range(1, 50, 2):
        odd.append(i)

    return odd

print(odd())
```

# 문항 2 반복문으로 네모 출력

```python
def rectangle(n, m):
    i = 0
    while i < m:
        print("*" * n)
        i += 1
    return None

rectangle(5,9)
```
