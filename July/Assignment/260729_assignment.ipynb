# 문항 2 Circle 클래스 생성, 활용

```python
class Circle:
    pi = 3.14

    def __init__(self, radius, x, y):
        self.radius = radius
        self.x = x
        self.y = y

    def area(self):
        return Circle.pi * (self.radius ** 2)

    def circumference(self):
        return Circle.pi * (self.radius * 2)

    def center(self):
        return self.x, self.y
```

```python
c1 = Circle(3, 2, 4)

print(c1.area()) #=> 28.26
print(c1.circumference()) #=> 18.84
print(c1.center()) #=> (2, 4)
```
