# 추가학습

## 도형 만들기

```python
class Point:
    x = 0
    y = 0
    def __init__(self, x: int, y: int):
        self.x = x
        self.y = y
```

```python
class Rectangle:
    def __init__(self, p1: Point, p2: Point):
        self.p1 = p1
        self.p2 = p2

    def get_area(self):
        height = abs(self.p1.y - self.p2.y)
        width = abs(self.p1.x - self.p2.x)
        return height * width

    def get_perimeter(self):
        height = abs(self.p1.y - self.p2.y)
        width = abs(self.p1.x - self.p2.x)
        return 2 * (height + width)

    def is_square(self):
        height = abs(self.p1.y - self.p2.y)
        width = abs(self.p1.x - self.p2.x)
        return True if height == width else False
```

```python
p1 = Point(1, 3)
p2 = Point(3, 1)
r1 = Rectangle(p1, p2)
print(r1.get_area())
print(r1.get_perimeter())
print(r1.is_square())

p3 = Point(3, 7)
p4 = Point(6, 4)
r2 = Rectangle(p3, p4)
print(r2.get_area())
print(r2.get_perimeter())
print(r2.is_square())
```
