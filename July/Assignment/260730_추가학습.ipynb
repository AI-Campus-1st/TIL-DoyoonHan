## 4. Seeding the Generator

```python
import random

print(random.random())
print(random.random())

random.seed(7777)
print(random.random())

random.seed(7777)
print(random.random())
```

```python
fake = Faker('ko_KR')
Faker.seed(4321)

print(fake.name())  # 1

fake2 = Faker('ko_KR')
print(fake2.name()) # 2
```

#1 과 #2의 값이 다르다. 그러므로 seed()는 classmethod이다.

```python
fake = Faker('ko_KR')
fake.seed_instance(4321)

print(fake.name())

fake2 = Faker('ko_KR')
print(fake2.name())
```

seed_instance()는 instancemethod이다. 

seed()는 여러 fake instance를 만들어도 다 같은 seed값을 사용하고싶을 때 사용하고,

seed_instance()는 여러 instance들이 전부 다른 seed값을 가지고 싶게 하고 싶을 때 사용한다.
