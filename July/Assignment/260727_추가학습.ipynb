# 추가학습

## 1. 세로로 출력하기

```python
num = int(input())

i = 0
while i < num:
    print(i + 1)
    i += 1
```

## 2. 거꾸로 세로로 출력하기

```python
num = int(input())

while num >= 0:
    print(num)
    num -= 1
```

## 3. N줄 덧셈

```python
num = int(input())

if num <= 10000:
    total = 0

    i = 1
    while i <= num:
        total += i
        i += 1

    print(total)
```

## 4. 간단한 N의 약수

```python
num = int(input())

if num <= 1000 and num >= 1:
    numbers = ""

    i = 1
    while i < num + 1:
        if num%i == 0:
            numbers += str(i)
            numbers += " "
        i += 1

    print(numbers.rstrip())
    
```

## 6. 계단 만들기

```python
def stairs():
    n = int(input())
    i = 1

    while i < n + 1:
        j = 1
        level = ""
        while j <= i:
            level += str(j)
            level += " "
            j += 1

        print(level.rstrip())
        i += 1

stairs()
```
