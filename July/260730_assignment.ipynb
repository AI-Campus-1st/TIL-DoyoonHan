# 문항 1 Dog과 Bird는 Animal이다

```python
class Animal:
	def __init__(self, name):
		self.name = name
	
	def walk(self):
		print(f'{self.name}! 걷는다!')
	
	def eat(self):
		print(f'{self.name}! 먹는다!')
```

```python
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    def walk(self):
        print(f'{self.name}! 딜린다!')

    def bark(self):
        print(f'{self.name}! 짖는다!')


class Bird(Animal):
    def __init__(self, name):
        super().__init__(name)

    def fly(self):
        print(f'{self.name}! 푸드덕!')
```

```python
dog = Dog('멍멍이')
print(dog.walk()) # 멍멍이! 달린다!
print(dog.bark()) # 멍멍이! 짖는다!

bird = Bird('구구')
print(bird.walk()) # 구구! 걷는다!
print(bird.eat()) # 구구! 먹는다!
print(bird.fly()) # 구구! 푸드덕!
```

# 문항 2 결제 시스템 구현

```python
class Payment:
    def __init__(self, owner: str, balance):
        self.owner = owner
        self.balance = balance

    def fee(self):
        pass

    def pay(self, amount):
        total = amount + self.fee(amount)
        if total > self.balance:
            raise ValueError("잔액 부족")
        else:
            self.balance -= total
            return {"amount" : amount, "fee": self.fee(amount), "balance": self.balance}
```

```python
class CardPayment(Payment):
    def __init__(self, owner, balance):
        super().__init__(owner, balance)
    
    def fee(self, amount):
        return int(amount * (2.5/100))

class MobilePayment(Payment):
    def __init__(self, owner, balance):
            super().__init__(owner, balance)
        
    def fee(self, amount):
        return 100

class PointPayment(Payment):
    def __init__(self, owner, balance):
            super().__init__(owner, balance)
        
    def fee(self, amount):
        return 0

    def pay(self, amount):
        try:
            dictionary = super().pay(amount)
            dictionary.update({"point_used": True})
            return dictionary
        except ValueError:
            raise ValueError("잔액 부족")
```

```python
# 실행 코드
print(CardPayment("홍길동", 20000).pay(10000))    # 수수료 250
print(MobilePayment("김철수", 20000).pay(10000))  # 수수료 100
print(PointPayment("이영희", 20000).pay(10000))   # 수수료 0, point_used True

print("---")
try:
    CardPayment("최가난", 5000).pay(10000)         # 잔액 부족
except ValueError as e:
    print(f"결제 실패: {e}")
```
