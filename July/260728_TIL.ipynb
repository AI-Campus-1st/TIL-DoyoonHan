## Built-in Functions

* map(func, iterable) -> return map object that is iterable that went through the func

* filter(func, iterable) -> return filter object with values that are only True from func

## Lambda

## Function

* default -> 기본 인자값을 가지는 인자 다음에 기본 값이 없는 인자를 사용할 수 없다.

* *args -> 임의의 개수의 위치인자를 받음. args로 func안에서 사용가능.

* *kwargs -> keyword arguments. keyword를 받을 수 있고, dict로 만듬.
