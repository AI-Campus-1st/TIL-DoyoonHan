## 시각화 설계 원칙

### 지각 정확도 순위

1. 공통 축 위의 위치 -> 막대 점 선

2. 축이 다른 위치 -> 소형 다중 차트

3. 길이 -> 누적 막대의 각 조각

4. 기울기 각도 -> 파이 차트 레이더

5. 면적 -> 버블 트리맵

6. 부피 색상 농도 -> 히트맵 3D

### Data-Ink Ratio

* 간결하게 만들어 요소를 줄인다. 

### 오해를 낳는 시각화

1. 축
    - y축 절단 막대 차트에서 하지 말것.
    - 이중 y축 가급적 회피
    - 면적 왜곡 -> 지름이 아니라 면적을 건드려야 한다.

2. 비교
    - 비교 기준이 잘못되면 결론이 뒤집힌다.
    1. 보정 없는 절대값
    2. 부적절한 기준

3. 색
    - 색은 역할이 있을 때만
    1. 범주형 -> 서로 다른 종류를 구분
    2. 순차형 -> 크기 강도의 많고 적음
    3. 발산형 -> 중앙 기존값 대비 양방향
