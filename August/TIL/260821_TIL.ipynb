* robots.txt 확인

```py
from urllib.robotparser import RobotFileParser
rp = RobotFileParser()
rp.set_url("https://example.com/robots.txt")
rp.read()
rp.can_fetch("*", "https://example.com/api/items")
# True / False
rp.crawl_delay("*") # 권고대기시간
```
