## Numpy 

(Numerical Python) 

```python
import numpy as np
```

* np.arange(시작 값, 끝값, step) -> python range()

```python
x = np.arange(10)
print(x)
x = np.arange(10.0)
print(x)
x = np.arange(1, 10, 2)
print(x)
x = np.arange(1, 10, 0.5)
print(x)
x = np.arange(10, 1, -0.5)
print(x)
```

* np.linspace(시작 값, 끝값, 벡터 크기) -> 균일한 간격으로 리스트 크기만큼의 리스트 생성

endpoint = False -> 끝 값 포함여부 옵션

```python
x = np.linspace(1, 20, 5)
print(x)
x = np.linspace(1, 20, 10, endpoint=True)
print(x)
```

* np.reshape(차원) -> 데이터를 유지하면서 차원의 형태를 변경

-1 일 경우 자동으로 맞춰 생성

```python
x = np.arange(9).reshape(3, 3)
print(x)
x = np.arange(24).reshape(2, 3, 4)
print(x)
x = np.array([[1, 2], [3, 4], [5, 6]])
print(x.reshape(2, 3))
x = np.arange(9).reshape(3, -1)
print(x)
```

* slicing

```python
x = np.arange(20)
print(x[1:3])
x = np.arange(20).reshape(4, 5)
print(x)
print(x[1:3])
print(x[1:3, 1:3])
x = np.arange(30).reshape(2, 5, 3)
print(x)
print(x[:, 3:5, 1])
```

* indexing & boolean indexing

```python
x = np.arange(20).reshape(4, 5)
print(x[1][1])
print(x[1, 1])
```

* random 

np.random.rand(5, 5) -> 난수가 들어가있는 다양한 형태의 데이터

np.random.randint(1, 10) -> 정수형 난수

np.random.randint(1, 10, size = (5)) -> 정수형 난수 크기 지정

* 특별한 형태의 배열

np.ones -> 모든 값이 1

np.zeroes -> 모든 값이 0

np.eyes(5) -> 단위행렬

ndarray.ravel(order = "C") -> 행렬 펼치기, F -> 열기준

* concatenate -> 배열을 연결

np.concatenate([x, y])

axis = 1 -> 열 기준으로 연결

```python
x = np.arange(1, 4)
y = np.arange(4, 7)
np.concatenate([x, y])

x = np.arange(10).reshape(2, 5)
y = np.arange(10, 20).reshape(2, 5)
np.concatenate([x, y])
np.concatenate([x, y], axis=1)
```

* split -> 배열을 분해

np.split(x, n)

axis = 1 -> 열을 기준으로 분해


```python
x = np.arange(12)
np.split(x, 4)
x = np.arange(16).reshape(4, 4)
np.split(x, 2, axis=1)
```

* broadcast -> 차원이 다른 행렬끼리의 계산

### functions

x = np.arange(4).reshape(2, 2)

y = np.arange(4).reshape(2, 2)

x.dot(y) # 행렬곱

np.transpose(x) # 전치행렬

np.linalg.inv(x) # 역행렬

np.linalg.det(x) # 행렬식

np.mean(x) # 평균

np.median(x) # 중간값

np.std(x) # 표준편차

np.var(x) # 분산

np.sum(x) #

np.sum(데이터, axis=1) #합, 축 변경

np.cumsum(x) # 누적합

np.cumprod(x) # 누적곱

np.min(x) # 최소값

np.argmin(x) #최소값 위치

np.argmax(x) # 최대값 위치

np.any(x > 4) # 하나라도 참이어야 참

np.all(x > 4) # 모든 요소가 참이어야 참

np.where(x > 4) # 조건에 맞는 위치
