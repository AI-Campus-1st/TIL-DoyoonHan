# 260811 Today I Learnt

### EXPLAIN

* type -> 테이블 접근 방식
    - ALL 이면 풀스캔중

* key -> 실제 선택된 인덱스
    - NULL이면 인덱스 미사용

* rows -> 읽을 것으로 추정한 행 수
    - 클 수록 의심

* Extra -> 추가로 일어나는 작업
    - filesort - temporary 주의

### ANALYZE

* EXPLAIN과 다르게 실젤 실행해봄.

* r_table_time_ms -> 테이블 데이터와 인덱스를 읽는 소요시간

* r_other_time_ms -> 테이블 생성, 정렬 등 기타 가공 처리 소요시간

* r_total_time_ms -> 전체 데이터 조회 소요시간

#### 진단 순서

1. 느린 쿼리 특정 -> 슬로우 쿼리 로그 또는 사용 경험

2. EXPLAIN -> type이 ALL인가, key가 NULL인가

3. ANALYZE -> r_rows와 r_total_time_ms로 진짜 병목 단계를 찾는다

4. 가설세우기 -> 인덱스 부재 · 함수 적용 · 조인 순서 · 통계 오류

5. 고친다 -> 인덱스 추가 또는 쿼리 재작성

6. 다시 잰다 -> ANALYZE로 전후를 수치로 비교한다

7. 부작용확인 -> 쓰기 성능과 디스크 사용량도 함께 본다


### ERD

#### 왜 계층을 나누는가

1. 재적재 가능성 -> 정제 로직이 틀려도 원본이 남아 있으면 다시 만들 수 있다

2. 장애 추적 -> 숫자가 이상할 때 어느 단계에서 틀어졌는지 좁힐 수 있다

3. 역할 분리 -> 적재/정제/분석 담당이 서로의 코드를 건드리지 않음

4. 성능 격리 -> 대시보드 조회가 원본 적재 작업과 자원을 다투지 않음
