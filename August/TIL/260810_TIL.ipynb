## SELECT

### WHERE

* condition
    - = <>
    - < > <= >=
    - AND OR NOT
    - IS NULL

* LIKE
    - '%a'
    - 'a%'
    - '%a%'

* IN
    - (1, 2, ...)
    - Basically a lot of ORs combined
    - NOT IN does not return NULL
    
* BETWEEN 1 AND 2
    - \>= 1 AND <= 2

* IFNULL / COALESCE
    - IFNULL(1, 'a')
    - COALESCE(1, 2, 'a')
