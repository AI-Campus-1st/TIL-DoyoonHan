### SQL

* DDL -> 데이터베이스, 테이블 같은 구조를 만들고 바꾼다
    - CREATE
    - ALTER
    - DROP
    - TRUNCATE
* DML -> 데이터를 추가하고, 조회하고, 수정하고, 삭제한다
    - INSERT
    - SELECT
    - UPDATE
    - DELETE
* DCL -> 누가 무엇을 할 수있는지 권한을 부여하고, 회수한다
    - GRANT
    - REVOKE
* TCL -> 변경을 확정하거나 되돌린다
    - COMMIT
    - ROLLBACK


### ACID

* Atomicity
    - 전부 반영되거나 전부 취소된다
* Consistency
    - 데이터는 미리 정의된 규칙에 의해서만 수정이 가능하다
* Isolation
    - 한 트랜잭션을 수행할 때, 다른 트랜잭션은 끼어들지 못한다
* Durability
    - 성공적으로 수행된 트랜잭션은 영원히 반영된다

localhost -> 이 PC에서만 접속 가능. % -> 어디서든 접속 가능. 192.168.0.10 -> 이 IP에서만 접속 가능

### 권환

* 모든 권한
    - GRANT ALL PRIVILEGES ON db.* TO 'user'@'localhost';

* 읽기 전용 계정
    - GRANT SELECT ON db.* TO 'viewer'@'localhost';

* 확인 / 회수 / 반영
    - SHOW GRANTS FOR 'user'@'localhost';
    - REVOKE SELECT ON db.* FROM 'viewer'@'localhost';
    - FLUSH PRIVILEGES;
