## 크롤러가 죽는 이유

* 그냥 실패할 확률도 있음.

* 시간이 길어져서

### 오류 type

1. ConnectionError
    - 서버/네트워크가 끊김

2. Timeout
    - 응답이 오지 않음 

3. 429 Too Many Requests
    - 속도를 줄이라는 신호

4. AttributeError:NoneType
    - 그 페이지에만 없는 항목

5. UnicodeDecodeError
    - 인코딩이 다른 페이지가 섞여 있음.
