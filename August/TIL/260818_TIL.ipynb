## HTTP

* HyperText Transfer Protocol

### 요청

1. 요청라인 Request Line (GET/POST)
    - Get: URL에 싣는다 
        - ?뒤에 변수 -> Query Parameter
    - Post: Body에 싣는다
        - 주소에 드러나지 않는 자리에 값을 담음

2. 헤더 Header

3. 바디 Body

### 응답

1. 상태줄 Status Line

2. 헤더 Header

3. 바디 Body

### 상태코드

* 2xx: 성공

* 3xx: 다른 곳으로

* 4xx: 요청이 잘몸ㅅ됨

* 5xx: 서버 문제

## 속성 선택자

1. 정확히 일치
    - a[target="_blank"]

2. 해당 단어를 포함
    - div[class~="active"]

3. 해당 값으로 시작
    - a[href^="/detail"]

4. 해당 값으로 끝
    - a[href$="ref=public-recruit"]

5. 해당 값을 포함
    - img[src*="cover200"]
