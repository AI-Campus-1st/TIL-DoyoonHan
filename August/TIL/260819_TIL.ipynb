## 정규표현식

### re.IGNORECASE
    - 케이스 구분하기 싫을 때 사용

### ^
    - 문자열의 시작
    - ^I like

### $
    - 문자열의 끝
    - orange!$

### .
    - 모든문자 하나
    - .... 하면 네 글자씩 묶어 추출

### []
    - 범위 지정
    - [orn] o, r, n 한글자
    - [0-9] 숫자 한글자
    - [A-Za-z] 영어 한글자
    - [가-힣] 한글 한글자

### [^]
    - 부정
    - [^0-9] 숫자가 아닌 것
    - <[^>]+ 태그 지울 때 쓰는 패턴

### ()
    - 묶음
    - | 사용해서 or 가능.

### 수량자
    - * 0번 이상
    - + 1번 이상
    - ? 0번 또는 1번

### {}
    - 개수 못박기
    - {5} 다섯글자

### 축약형
    - \d -> [0-9]
    - \w -> [a-zA-Z0-9_]
    - \s -> 공백/탭/줄바꿈
    - 반대로 하고 싶으면 UpperCase

### ?=
    - 전방 탐색
    - 뒤에 무엇이 오는지를 조건으로, 결과에는 넣지않음.
    - oran(?=ge!) -> oran 을 가져옴

### ?<=
    - 후방 탐색
    - 앞에 무엇이 있는지가 조건

### ?!
    - 부정형

## 자주 재사용 할 수 있는 함수

```py
import re, html

def clean_text(s):
    s = re.sub(r"<[^>]+>", "", s) # 남은태그
    s = html.unescape(s) # &nbsp; 같은엔티티
    s = re.sub(r"\s+", " ", s) # 연속공백
    s = re.sub(r"\[[^\]]*\]", "", s) # [ 단독] 말머리
    return s.strip()
```
