## 문항 1 할리스커피 매장 정보 10페이지 수집

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
URL = 'https://www.hollys.co.kr/store/korea/korStore2.do'
```

```python
extracted = []

for i in range(1, 11):
    params = {"pageNo" : i, 
              "sido" : "",
              "gugun" : "",
              "store" : ""}
    
    soup = BeautifulSoup(requests.get(URL, params=params, headers={"User-Agent": "Mozilla/5.0"}).text, 'html.parser')
    trs = soup.find("table", attrs={"class" :"tb_store"}).find("tbody").find_all("tr")

    for tr in trs:
        td = tr.select("td")

        # 매장 서비스
        icons = td[4].select("img")
        services = []
        for icon in icons:
            services.append(icon.attrs["alt"])

        extracted.append({
            '지역' : td[0].text.strip(),
            '매장명' : td[1].text.strip(),
            '현황' : td[2].text.strip(),
            '주소' : td[3].text.strip(),
            '매장 서비스' : services,
            '전화번호': td[5].text.strip()
        })

print(extracted)
```

```python
df = pd.DataFrame(extracted)
df.to_csv('hollys.csv', index=False, encoding='utf-8-sig')
```

## 문항 2 알라딘 베스트셀러 수집

```python
URL = 'https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1'
extraction = []

for i in range(1, 11):

    params = {
        "Page" : i,
    }

    soup = BeautifulSoup(requests.get(URL, params=params, headers={"User-Agent": "Mozilla/5.0"}).text, 'html.parser')
    books = soup.find_all("div", class_="ss_book_box")

    for book in books:
        extraction.append({
            "카테고리" : book.select_one("span.tit_category").text.strip(),
            "제목" : book.select_one("a.bo3").text.strip(),
            "저자" : book.select_one("li:nth-of-type(3)").text.strip(),
            "정가" : book.select_one("li:nth-of-type(4) > span").text.strip(),
            "활인가격" : book.select_one("span.ss_p2").text.strip(),
            "이미지" : book.select_one("img").attrs["src"],
        })
```

```python
df = pd.DataFrame(extraction)
df.to_csv('aladin_bestseller.csv', index=False, encoding='utf-8-sig')
```
