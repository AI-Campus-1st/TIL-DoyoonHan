# 문항 1 무신사 상품 10페이지

```python
import requests
import pandas as pd
import time

URL = "https://api.musinsa.com/api2/dp/v2/plp/goods"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
}

result = []

MAX_PAGE = 10
next_Page_Url = None
for page in range(1, MAX_PAGE+1):
    PARAMS = {
        'gf': 'A', 
        'sortCode': 'POPULAR', 
        'category': '001',
        'size': '60',
        'caller': 'CATEGORY',
        'seen': '121',
    }

    if next_Page_Url is None:
        response = requests.get(URL, params={**PARAMS, "page": page}, headers=HEADERS, timeout=10)
    else:
        response = requests.get(next_Page_Url, headers=HEADERS, timeout=10)

    products = response.json()
    for product in products['data']['list']:
        result.append({
            "브랜드명": product.get('brandName', 'None'),
            "제품명": product.get('goodsName', 'None'),
            "원래가격": product.get('normalPrice', 0),
            "할인가격": product.get('finalPrice', 0),
            "리뷰수": product.get('reviewCount', 0),
            "리뷰점수": product.get('reviewScore', 0),
        })

    next_Page_Url = products['data']['pagination']['nextPageUrl']
    time.sleep(1)

df = pd.DataFrame(result)
df.to_csv("musinsa.csv", index=False)
```

# 문항 2 로켓펀치 채용공고 10페이지

```python
import requests
import pandas as pd
import time

URL = "https://www.rocketpunch.com/api/proxy/jobs"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
    "x-rocket-app-key": "077cd9d8-7237-4bee-b2d3-77690d163cca",
    "x-rocket-client-type": "WEB",
    "x-rocket-device-type": "PC",
    "x-rocket-os-type": "WIN",
}
PARAMS = {
    "sort" : "DATE_DESC",
}
MAX_PAGE = 10

page_token = None
result = []

for page in range(1, MAX_PAGE + 1):
    response = requests.get(URL, params={**PARAMS, "pageToken": page_token},
                    headers=HEADERS, timeout=10)
    response.raise_for_status()
    data = response.json()
    for item in data['items']:
        result.append({
            "기업명": item.get('companyName', ''),
            "공고명": item.get('title', ''),
            "요약": item.get('description', ''),
            "업무형태": item.get('workType', ''),
        })
    page_token = data['pageToken']
    time.sleep(1)

df = pd.DataFrame(result)
df.to_csv("rocketpunch.csv", index=False)
```
